```python
# 観測値と分布のパラメータ設定
x = 2.0       # 観測値
sigma1 = 1.0   # 分散は既知とする（sigma^2）
sigma2 = 0.5   # 分散は既知とする（sigma^2）

# θ の範囲
theta = np.linspace(-10, 10, 400)

# 対数尤度（正規 N(θ, σ^2) のとき）
loglik1 = -0.5 * np.log(2 * np.pi * sigma1**2) - (x - theta)**2 / (2 * sigma1**2)
loglik2 = -0.5 * np.log(2 * np.pi * sigma2**2) - (x - theta)**2 / (2 * sigma2**2)

# プロット
plt.figure(figsize=(10, 6))
plt.plot(theta, loglik1, label='Log-Likelihood', color='blue')
plt.plot(theta, loglik2, label='Log-Likelihood (2*sigma^2)', color='red')
```
